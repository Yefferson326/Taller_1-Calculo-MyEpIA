## Parte 1: Visualización de la Superficie y Curvas de Nivel

Dada la función de dos variables:

$$f(x,y) = \frac{5xy}{x^2 + y^2 + 1}$$

Generamos una grilla bidimensional en el dominio $x \in [-4, 4]$ y $y \in [-4, 4]$ para evaluar la superficie y graficar sus curvas de nivel.

```python
# 1. Crear el dominio y la grilla bidimensional
x_vals = np.linspace(-4, 4, 300)
y_vals = np.linspace(-4, 4, 300)
X, Y = np.meshgrid(x_vals, y_vals)

# Evaluamos la función de manera vectorizada
Z = (5 * X * Y) / (X**2 + Y**2 + 1)

# 2. Visualización con plt.contour
plt.figure(figsize=(8, 6))
contour = plt.contour(X, Y, Z, levels=30, cmap='viridis')
plt.clabel(contour, inline=True, fontsize=8, fmt='%1.2f')

plt.title('Curvas de Nivel de $f(x,y) = \\frac{5xy}{x^2 + y^2 + 1}$', fontsize=12)
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, linestyle='--', alpha=0.5)
plt.axhline(0, color='black', linewidth=0.8, linestyle='--')
plt.axvline(0, color='black', linewidth=0.8, linestyle='--')
plt.colorbar(contour, label='$f(x,y)$')
plt.show()
```

### Pregunta Conceptual (Parte 1)
**¿Qué representa físicamente una curva de nivel cuando $f(x,y)$ mide el error de un modelo de IA?**

En el contexto de Inteligencia Artificial y Aprendizaje Automático, la función $f(x,y)$ actúa como una **Función de Pérdida (Loss Function)** o superficie de error, donde $x$ e $y$ representan dos parámetros ajustables del modelo (por ejemplo, pesos $w_1$ y $w_2$).

Una **curva de nivel** representa una **línea de equierror (o isolínea)**: es decir, el lugar geométrico de todas las combinaciones posibles de parámetros $(x,y)$ que generan **exactamente el mismo nivel de error** $C$ en el modelo ($f(x,y) = C$). 

- Moverse **a lo largo** de una curva de nivel significa modificar los pesos sin alterar el desempeño del modelo.
- Moverse de manera **perpendicular** a la curva de nivel implica cambiar el error al ritmo más acelerado posible (aumento o disminución).
